```python
from pathlib import Path

env_path = Path.cwd() / ".env"
if not env_path.exists():
    env_path.write_text("GOOGLE_API_KEY=\n")
env_path.resolve()
```

```python
# Smoke test (to run locally after install).
import os, sys, platform, importlib.util

print("Python:", sys.version)
print("Platform:", platform.platform())
print("GOOGLE_API_KEY set:", bool(os.environ.get("GOOGLE_API_KEY") or os.environ.get("GEMINI_API_KEY")))

# Check for google.genai module (provided by google-adk)
genai_spec = importlib.util.find_spec("google.genai")
print("google.genai module found:", bool(genai_spec))
if not genai_spec:
    print("If False, ensure your venv is active and 'google-adk' is installed in it.")
else:
    print("Success! You can now use: import google.genai as genai")
```
